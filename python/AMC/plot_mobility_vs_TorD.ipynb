```python
import numpy as np
import matplotlib.pyplot as plt
import glob, os
import pandas as pd

import scienceplots
plt.style.use(['science', 'notebook', 'grid'])
```

```python
folder = "../../build/FullMonaFullmainBulk/**/"
files = glob.glob(
    os.path.join(folder, "mobility_summary.txt")
)
print(f"Found {len(files)} files.")
print("File 0:", files[0])
```

```python
list_D = []
list_mobility = []

print(f"Extracting doping concentration and mobility from {len(files)} files...")
for file in files:
    D = float(file.split("/")[-2].split("_")[1][2:])
    list_D.append(D)
    print(f"Processing file: {file}, D={D:.2e} cm^-3")
    # Read the mobility from the file
    with open(file, "r") as f:
        lines = f.readlines()
        for line in lines:
            print(f"Line: {line.strip()}")
            if line.startswith("low_field_mobility_cm2_per_V_s"):
                mobility = float(line.split("=")[1].strip().split()[0])
                list_mobility.append(mobility)
                print(f"D={D:.2e} cm^-3, mobility={mobility:.2f} cm^2/V/s")
                break
            
# Convert lists to numpy arrays for plotting
D_array = np.array(list_D)
mobility_array = np.array(list_mobility)
argsort = np.argsort(D_array)
D_array = D_array[argsort]
mobility_array = mobility_array[argsort]

fig, ax = plt.subplots()
ax.plot(D_array, mobility_array, marker='o')
ax.set_xlabel("Doping concentration (cm$^{-3}$)")
ax.set_ylabel("Low-field mobility (cm$^2$/V/s)")
ax.set_title("Low-field mobility vs Doping concentration")
ax.set_xscale("log")
plt.show()
```

```python
folder = "../../build/BULKT/**/"
files = glob.glob(
    os.path.join(folder, "mobility_summary.txt")
)
print(f"Found {len(files)} files.")
print("File 0:", files[0])
```

```python
list_T = []
list_mobility = []
print(f"Extracting temperature and mobility from {len(files)} files...")
for file in files:
    T = float(file.split("/")[-2].split("_")[-1][:])
    list_T.append(T)
    print(f"Processing file: {file}, T={T:.2f} K")
    # Read the mobility from the file
    with open(file, "r") as f:
        lines = f.readlines()
        for line in lines:
            print(f"Line: {line.strip()}")
            if line.startswith("low_field_mobility_cm2_per_V_s"):
                mobility = float(line.split("=")[1].strip().split()[0])
                list_mobility.append(mobility)
                print(f"T={T:.2f} K, mobility={mobility:.2f} cm^2/V/s")
                break

# Convert lists to numpy arrays for plotting
T_array = np.array(list_T)
mobility_array = np.array(list_mobility)
argsort = np.argsort(T_array)
T_array = T_array[argsort]
mobility_array = mobility_array[argsort]

ref_file = "../../examples/references/Mobility_PureSi_vs_T_Canali.csv"
print(f"Reading reference data from {ref_file}...")
X,Y = np.loadtxt(ref_file, delimiter=",", skiprows=1, unpack=True)

fig, ax = plt.subplots()
ax.plot(T_array, mobility_array, marker='o', label="PBMC")
ax.plot(X, Y, label="Canali", marker='+')
ax.set_xlabel("Temperature (K)")
ax.set_ylabel("Low-field mobility (cm$^2$/V/s)")
ax.set_title("Low-field mobility vs Temperature")
ax.set_xscale("log")
ax.set_yscale("log")
ax.legend()
ax.grid(True, which="both", ls="--")
plt.show()
```

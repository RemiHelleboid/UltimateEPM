```python
import numpy as np
import matplotlib.pyplot as plt
import glob, os
import pandas as pd

import scienceplots
plt.style.use(['science', 'notebook', 'grid'])
```

```python
folder = "../build/mainBulk"
files = glob.glob(
    os.path.join(folder, "**/observables.csv")
)
print(f"Found {len(files)} files.")
```

```python
fig, axs = plt.subplots(1, 1, figsize=(10, 10), sharex=True)

def van_overstraeten_de_man_alpha_n(electric_field_V_per_m, temperature_K=300.0):
    electric_field_V_per_cm = np.asarray(electric_field_V_per_m) / 100.0

    a_inf = 7.03e5  # cm^-1
    b = 1.231e6     # V/cm

    hbar_omega_op_eV = 0.063
    k_B_eV_per_K = 8.617333262145e-5
    T0 = 300.0

    gamma = np.tanh(hbar_omega_op_eV / (2.0 * k_B_eV_per_K * T0)) / \
            np.tanh(hbar_omega_op_eV / (2.0 * k_B_eV_per_K * temperature_K))

    return gamma * a_inf * np.exp(-gamma * b / electric_field_V_per_cm)


list_electric_fields = []
list_mean_velocities = []
list_mean_energies = []
list_impact_ionization_coefficients = []

# electric_field_V_per_m,mean_velocity_x_m_per_s,mean_kinetic_energy_eV,sample_count,impact_ionization_events,impact_ionization_rate_per_carrier_s_1,impact_ionization_drift_velocity_m_per_s,impact_ionization_coefficient_cm_1


for file in files[:]:
    data = np.loadtxt(file, delimiter=",", skiprows=1)
    electric_field = data[0]
    print(f"Electric field: {electric_field} V/m")
    mean_velocity = data[1]
    mean_energy = data[2]
    impact_ionization_coefficient = data[-1]

    list_electric_fields.append(electric_field)
    list_mean_velocities.append(mean_velocity)
    list_mean_energies.append(mean_energy)
    list_impact_ionization_coefficients.append(impact_ionization_coefficient)

argsorted_indices = np.argsort(list_electric_fields)
list_electric_fields = np.array(list_electric_fields)[argsorted_indices]
list_mean_velocities = np.array(list_mean_velocities)[argsorted_indices]
list_mean_energies = np.array(list_mean_energies)[argsorted_indices]
list_impact_ionization_coefficients = np.array(list_impact_ionization_coefficients)[argsorted_indices]

axs.plot(list_electric_fields, list_mean_velocities, label="Mean velocity")

axs.set_ylabel("Mean velocity (m/s)")
axs.set_xlabel("Electric field (V/m)")
axs.legend()


fig, axs = plt.subplots(1, 1, figsize=(10, 10), sharex=True)
axs.plot(list_electric_fields, list_mean_energies, label="Mean kinetic energy")
axs.set_ylabel("Mean kinetic energy (eV)")
axs.set_xlabel("Electric field (V/m)")
axs.legend()


list_ef = np.linspace(min(list_electric_fields), max(list_electric_fields), 100)
list_alpha_van_overstraeten = van_overstraeten_de_man_alpha_n(list_ef)
fig, axs = plt.subplots(1, 1, figsize=(10, 10), sharex=True)
axs.plot(1e6*100.0/list_electric_fields, list_impact_ionization_coefficients, 'o', label="MC simulation")
axs.plot(1e6*100.0/list_ef, list_alpha_van_overstraeten, label="Van Overstraeten-De Man model")
axs.set_ylabel("Impact ionization coefficient (cm^-1)")
axs.set_xlabel("1 / Electric field (cm/V)")
axs.set_xlim(1, 7)
axs.set_ylim(ymin=1e2)
axs.set_yscale("log")
axs.legend()
```
